# 2.0.0 - Xây dựng biến lợi suất (Daily Log Return)

**Công thức tính toán:**
$$\text{Return}_{1D, t} = \ln\left(\frac{\text{Close}_t}{\text{Close}_{t-1}}\right)$$

- Dữ liệu đầu vào: `dataset/1.1.0_data_cleaning.csv`
- Dữ liệu đầu ra: `dataset/2.0.0_return.csv`

```python
import os
import numpy as np
import pandas as pd

# 1. Tự động định vị đường dẫn thư mục dataset
if os.path.exists('dataset'):
    dataset_dir = 'dataset'
elif os.path.exists('../dataset'):
    dataset_dir = '../dataset'
else:
    dataset_dir = '.'

input_path = os.path.join(dataset_dir, '1.1.0_data_cleaning.csv')
print(f"Đọc dữ liệu từ: {input_path}")
df = pd.read_csv(input_path)

# Sắp xếp theo ngày tăng dần để đảm bảo tính chuỗi thời gian
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date').reset_index(drop=True)
print(f"Kích thước dữ liệu đầu vào: {df.shape[0]} dòng, {df.shape[1]} cột.")
display(df.head(3))
```

```python
# 2. Xây dựng biến lợi suất 1 ngày (Daily Log Return)
# Công thức: Return_1D,t = ln(Close_t / Close_t-1)
df['Return_1D'] = np.log(df['Close'] / df['Close'].shift(1))

# Ngày đầu tiên (t=0): lấy ln(Close_0 / Open_0) để giữ nguyên 100% số dòng không bị mất
df.loc[0, 'Return_1D'] = np.log(df.loc[0, 'Close'] / df.loc[0, 'Open'])

print("5 dòng đầu tiên của dữ liệu (đã tính đủ cả dòng đầu tiên):")
display(df[['Date', 'Open', 'Close', 'Return_1D']].head())
```

```python
# 3. Kiểm tra thống kê mô tả
print(f"Tổng số dòng: {len(df)} (NaN = {df['Return_1D'].isnull().sum()})\n")
print("=== THỐNG KÊ MÔ TẢ BIẾN LỢI SUẤT (Return_1D) ===")
print(df['Return_1D'].describe())
```

```python
# 4. Lưu dữ liệu kết quả phiên bản 2.0.0
# Quy ước: File CSV xuất ra có cùng tên với file notebook (2.0.0_return.csv)
output_path = os.path.join(dataset_dir, "2.0.0_return.csv")
df.to_csv(output_path, index=False)

print(f"✅ ĐÃ XUẤT THÀNH CÔNG TẠI: {output_path}")
print("\n5 dòng đầu của dữ liệu mới:")
display(df.head(3))
```
